````python
# First neoron
import numpy as np
def neuron(inputs,weights,bias,activationfunction="relu"):

  # Step 1: Weight Sum
  z=np.dot(inputs,weights)+b
  # Step 2: Activation Function
  if activationfunction=="relu":
    return np.maximum(0,z)
  elif activationfunction=="sigmoid":
    return 1/(1+np.exp(-z))
  elif activationfunction=="tanh":
    return np.tanh(z)
  return z
inputs=np.array([100,35,24])
weights=np.array([0.35,0.50,-10])
b=100
output=neuron(inputs,weights,b,activationfunction="sigmoid")
print(output)
````

## Activation Function

Activation Function is just a mathematical bend we apply to neuron output. or A nonlinear function applied to a neuron's weighted sum.

**Without Activation**: [Neuron] → Straight Line

**With Activation**: [Neuron] → Curved output


### Four Activation Functions

#### 1. Sigmoid Function
- **Formula**: f(x) = 1 / (1 + e^(-x))
- **Range**: 0 to 1
- **Use**: Binary classification (output layer)
- **Pros**: Smooth gradient, outputs probabilities
- **Cons**: Vanishing gradient, not zero-centered

#### 2. Tanh (Hyperbolic Tangent)
- **Formula**: f(x) = (e^x - e^(-x)) / (e^x + e^(-x))
- **Range**: -1 to 1
- **Use**: Hidden layers
- **Pros**: Zero-centered, stronger gradients
- **Cons**: Still suffers from vanishing gradient

#### 3. ReLU (Rectified Linear Unit)
- **Formula**: f(x) = max(0, x)
- **Range**: 0 to infinity
- **Use**: Hidden layers (most popular)
- **Pros**: Fast, sparse activation, no vanishing gradient
- **Cons**: "Dying ReLU" (neurons can die for x < 0)

#### 4. Softmax
- **Formula**: f(x) = e^x / Σe^x
- **Range**: 0 to 1 (sums to 1)
- **Use**: Multi-class classification (output layer)
- **Pros**: Outputs probabilities, sums to 1
- **Cons**: Sensitive to outliers


### Quick Comparison

| Function | Formula | Range | Best For |
|----------|---------|-------|----------|
| Sigmoid | 1/(1+e^(-x)) | (0, 1) | Binary classification |
| Tanh | tanh(x) | (-1, 1) | Hidden layers |
| ReLU | max(0, x) | [0, ∞) | Hidden layers |
| Softmax | e^x/Σe^x | (0, 1) | Multi-class |

````python
# Sigmoid
import matplotlib.pyplot as plt
import numpy as np

def sigmoid(z):
  return 1/(1+np.exp(-z))

# Test it
X=np.linspace(-10,10,100)
Y=sigmoid(X)
# Plotting
plt.figure(figsize=(8,6))
plt.plot(X,Y,color="green")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Sigmoid Smashes to [0,1]")
plt.axhline(y=1,color="grey",linestyle="--")
plt.axhline(y=0,color="grey",linestyle="--")
plt.axvline(x=0,color="grey",linestyle="--")
plt.grid(True)
plt.text(-5,0.5,"Zero-centered\ngood for probabilities\n output is always 0-1",fontsize=10)
plt.show()
````

````python
# Tanh
def tanh(x):
  return np.tanh(x)
X=np.linspace(-10,10,100)
Y=tanh(X)

plt.figure(figsize=(8,6))
plt.plot(X,Y,color="red")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Tanh Smashes to [0,1]")
plt.axhline(y=-1,color="grey",linestyle="--")
plt.axhline(y=1,color="grey",linestyle="--")
plt.axvline(x=0,color="grey",linestyle="--")
plt.grid(True)
plt.text(-5,0,"Zero-centered\ngood for hidden layers",fontsize=10)
````

````python
# relu
def relu(x):
  return np.maximum(0,x)
X=np.linspace(-10,10,100)
Y=relu(X)

plt.figure(figsize=(8,6))
plt.plot(X,Y,color="blue")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Relu Smashes to [0,1]")
plt.axhline(y=0,color="grey",linestyle="--")
plt.axvline(x=0,color="grey",linestyle="--")
plt.grid(True)
plt.text(-5,0,"most popular today and \n veray fast smooth \n and good works",fontsize=10)
plt.show()
````

````python
# softmax
def softmax(x):
   exp_x = np.exp(x - np.max(x)) # stable: subtract max
   return exp_x / exp_x.sum(axis=0)
X=np.linspace(-10,10,100)
Y=softmax(X)
plt.figure(figsize=(8,6))
plt.plot(X,Y,color="orange")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Softmax Smashes to [0,1]")
plt.grid(True)
plt.axhline(y=0,color="grey",linestyle="--")
plt.axhline(y=1,color="grey",linestyle="--")
plt.axvline(x=0,color="grey",linestyle="--")
plt.show()
````

## Network Architecture
Neural networks are connected into layer. **Layer** is a collection of neurons that transfer data from one layer to other.

> There are different types of layers.

**1. Input Layer :** Receives dat passes raw inputs forward no mathamatical computation happens here.

**2. Hidden Layer :** Recives the input from input layer and then playes the mathamatical functions and learn patterns , combination dataset feature and abstraction.

**3. Output Layer :** Produces final result and final transfromation happens here to generate the predictions or ansers.

**Key points:**
1. Shallow networks: 1-2 hidden layers
2. Deep networks: 3+ hidden layers and this is called deep learning.
3. Width: Number of neurons in a single layer.

### How information flows.
##### **1. Forwared Propagation:**
Forward propagation is the process of moving data from input to output through the Network .Each layer recives ouput of previous layer . Computes a weighted sum and applies activation function before passig result forward.

**Forward propagation step by step :** 
1. Step 1: Input feature enter the network e.g: [2.5,1.8,0.3]
2. Step 2: Each hidden layer(neuron) recives all inputs , multiplies with weights, adds bias and applies activation function.
3. Step 3: outputs of one layers becomes input to the next layer.
4. Step 4: Repeat until final output layer produces the prediction.


````python
import numpy as np
# Hands-on: Forward Propagation from Scratch
# 1. Inputs
x1 = 2
x2 = 3
# 2. Hidden neuron parameters
w1 = 0.5
w2 = 0.3
b1 = -1
# 3. Output neuron parameters
w_out = 0.8
b_out = 0
# Step 1: Hidden Neuron
Z = (w1 * x1) + (w2 * x2) + b1
print(f"Z: {Z:.2f}")
# Step 2: Apply ReLU
relu = max(0, Z)
print(f"ReLU(Z): {relu:.2f}")
# Step 3: Output Neuron
Z_out = (relu * w_out) + b_out
print(f"Z_out: {Z_out:.2f}")
# Step 4: Apply Sigmoid
sigmoid = 1 / (1 + np.exp(-Z_out))
print(f"Sigmoid: {sigmoid:.4f}")
print(f"{sigmoid * 100:.2f}% probability of Positive Class.")
````
